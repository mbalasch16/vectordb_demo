```python
from langchain_community.vectorstores import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
```

```python
loader = DirectoryLoader("./data/",glob="./*.txt",loader_cls=TextLoader)
textLoader = loader.load()
textLoader
```

```python
import langchain
print(langchain.__version__)
```

```python
# Import a commonly used text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Extract the text from the Document objects
#text = " ".join([doc.page_content for doc in textLoader])
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=50)
chunks = splitter.split_documents(textLoader)

print(chunks)
```

creating db

```python

from langchain_huggingface import HuggingFaceEmbeddings

persist_directory = "db"

embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vectordb = Chroma.from_documents(documents = chunks, embedding = embedding, persist_directory=persist_directory)
```
